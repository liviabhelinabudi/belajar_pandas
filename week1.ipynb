Materi 2: Pemilihan Baris (Rows Selection) pada Data Frame

Import Modules

```python
import pandas as pd
import numpy as np

print(pd.__version__)
print(np.__version__)
```

Persiapan Data Frame

```python
n_rows = 10
n_cols = 5
cols = tuple('ABCDE')

df = pd.DataFrame(np.random.randint(1, 5, size=(n_rows, n_cols)), 
                  columns=cols)
df
```

Selection dengan Operator Logika |

```python
df[(df['A'] == 1) | (df['A'] == 3)]
```

Selection dengan Fungsi isin()

```python
df[df['A'].isin([1, 3])]
```

Mengenal Operator Negasi ~


```python
df[~df['A'].isin([1, 3])]
```

Materi 3: Konversi Tipe Data String ke Numerik pada Kolom Data Frame

Import Modules

```python
import pandas as pd

print(pd.__version__)
```

Persiapan Data Frame

```python
data = {'col1':['1', '2', '3', 'teks'], 
        'col2':['1', '2', '3', '4']}

df = pd.DataFrame(data)
df
```

```python
df.dtypes
```

Konversi Tipe Data dengan Fungsi astype()


```python
df_x = df.astype({'col2':'int'})
df_x
```

```python
df_x.dtypes
```

Konversi Tipe Data Numerik dengan Fungsi to_numeric()

```python
df.apply(pd.to_numeric, errors='coerce')
```

Materi 4: Pemilihan Kolom (Columns Selection) pada Data Frame berdasarkan Tipe Data


Import Modules


```python
import pandas as pd
import numpy as np

print(pd.__version__)
print(np.__version__)
```

Persiapan Data Frame


```python
n_rows = 5
n_cols = 2
cols = ['bil_pecahan', 'bil_bulat']

df = pd.DataFrame(np.random.randint(1, 20, size=(n_rows, n_cols)), columns=cols)

df['bil_pecahan'] = df['bil_pecahan'].astype('float')

df.index = pd.date_range(start='2024-10-01', periods=n_rows, freq='h')

df = df.reset_index()
df['teks'] = list('ABCDE')

df
```

```python
df.dtypes
```

Memilih Kolom Bertipe Data Numerik


```python
df.select_dtypes(include='number')
```

```python
df.select_dtypes(include='float')
```

```python
df.select_dtypes(include='int')
```

Memilih Kolom Bertipe Data string atau object


```python
df.select_dtypes(include='object')
```

Memilih Kolom Bertipe Data datetime

```python
df.select_dtypes(include='datetime')
```

Memilih Kolom dengan Kombinasi Tipe Data

```python
df.select_dtypes(include=['number', 'object'])
```
