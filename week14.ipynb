Materi 53: Membuat kategori baru berdasarkan threshold (ambang batas)

Import Modules

```python
import pandas as pd

print(pd.__version__)
```

Persiapan Data Frame

```python
d = {'hobi':['jogging', 'mancing', 'renang',
             'mancing', 'mancing', 'baca', 
             'baca', 'mancing', 'fotograsi', 
             'mancing', 'camping']}

df = pd.DataFrame(d)
df
```

Membuat kategori baru berdasarkan threshold (ambang batas)

```python
df['hobi'].value_counts()
```

```python
persentase = df['hobi'].value_counts(normalize=True)
persentase
```

```python
threshold = 0.1
hobi_lain = persentase[persentase < threshold].index
hobi_lain
```

```python
df['hobi'] = df['hobi'].replace(hobi_lain, 'lainnya')
df['hobi']
```

```python
df['hobi'].value_counts(normalize=True)
```

Materi 55: Menghitung jumlah baris menurut kriteria tertentu

Import Modules

```python
import pandas as pd
import numpy as np

print(pd.__version__)
print(np.__version__)
```

Persiapan Data Frame

```python
n_rows = 10
n_cols = 5
cols = tuple('ABCDE')

df = pd.DataFrame(np.random.randint(1,10, size=(n_rows, n_cols)),
                    columns=cols)
df
```

Menghitung jumlah baris menurut kriteria tertentu

```python
(df['A'] < 5).sum()
```
